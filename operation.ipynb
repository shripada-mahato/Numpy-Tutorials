Print the original array.
Print the original dtype.
Convert the array to int using astype().
Store it in a new variable named new_arr.
Print:
new_arr
new_arr.dtype
Finally, print the original array again.

```python
import numpy as np
arr = np.array([10.5, 20.7, 30.1, 40.9])
print("original array:- \n", arr)
print(arr.dtype)
new_int_arr= arr.astype(int)
print(new_int_arr,"\n",new_int_arr.dtype)
print(arr)
```

Write a program that:

Creates a 2 × 4 array using np.empty().
Prints:
the array
ndim
shape
size
dtype
Then replace every element with 7 using slicing.
Print the array again.

```python
import numpy as np
arr = np.empty((2,4))
print("array:- \n",arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
arr[:]=7
print("new array:- \n",arr)
```
