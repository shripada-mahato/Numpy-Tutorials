Print the original array.
Print the original dtype.
Convert the array to int using astype().
Store it in a new variable named new_arr.
Print:
new_arr
new_arr.dtype
Finally, print the original array again.

```python
import numpy as np
arr = np.array([10.5, 20.7, 30.1, 40.9])
print("original array:- \n", arr)
print(arr.dtype)
new_int_arr= arr.astype(int)
print(new_int_arr,"\n",new_int_arr.dtype)
print(arr)

```
